####Iterators

Iterators are advanced Python concept that allow for efficient looping and memory management

```python
my_list = [1,2,3,4,5,6]

for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

```python
####Iterator

iterator = iter(my_list)
print(type(iterator))
```

```python
print(iterator)
```

```python
#### iterate through all the element
next(iterator)
```

```python
iterator = iter(my_list)
```

```python

try:
    print(next(iterator))
except StopIteration:
    print("There are no element in the iterator")
```

```python
str="hello"
str_iterator = iter(str)
```

```python

try:
    print(next(str_iterator))
except StopIteration:
    print("There are no element in the str_iterator")
```

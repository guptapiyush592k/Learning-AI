## Reading data from different Source

```python
import pandas as pd
from io import StringIO

data = '{"id": 101,"first_name": "Piyush","last_name": "Gupta","email": "piyush.gupta@example.com","age": 27,"gender": "Male","designation": "Frontend Developer","manager": [{"id": 12,"name": "Rahul Sharma"}]}'

df = pd.read_json(StringIO(data))
```

```python
df
```

```python
df.to_json()
```

```python
df.to_json(orient='index')
```

```python
df.to_json(orient='records')
```

## Pandas

Its a powerfull data manupulation library, widely used for data analysis and data cleaning. it provides two primary data structures: series and Dataframe. A Series is 1D array like object, while a DataFrame is 2D, size-mutable  and potential heterogeneous tabulor data structure with labeled axes(row and columns).

```python
import pandas as pd
```

```python
import pandas as pd

data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
print(type(series))
```

```python
#series from dict obj
data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index = ['a', 'b', 'c']

pd.Series(data, index)
```

```python
## DataFrame

## Create a DataFrame from a dict of list

data = {
    'name':['krish', 'piyush', 'john'],
    'age':[25,26,30],
    'city':['banglore', 'bareilly', 'delhi']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## create a dataframe from a list of dict.

data = [
    {'name':'krish','age':32, 'city' : 'banglore'},
    {'name':'Jogn','age':33, 'city' : 'Pune'},
    {'name':'Raj','age':34, 'city' : 'Lucknow'}
]

df1 = pd.DataFrame(data)
print(df1)
```

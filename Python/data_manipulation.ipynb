##  Data Manipulation and Analysis with Pandas

It is key task in any data science or data analysis project. Pandas provide a wide range of functions for data manipulation and analysis, making it clean, transform, and extract insights from data.


```python
import pandas as pd
print("hi")
```

```python
df = pd.read_csv('dummy.csv')
df.head(5)
```

```python
## Handling missing value
df.isnull()
```

```python
df.isnull().any(axis=1)  # it means row wise tell where data is missing
```

```python
df.isnull().any() #it will tell which column has missing value
```

```python
df.isnull().sum() #is telling which column has how many missing value
```

```python
df_filled = df.fillna(0) #replace all the missing value with 0
df_filled
```

```python
## filling missing value with the mean of the column

df['sales_fillNA'] = df['monthly_sales'].fillna(df['monthly_sales'].mean())
df.head(10)
```

```python
## Renaming columns
df = df.rename(columns={'salary':'CTC'})
df.head()
```

```python
##Change dataTypes

df['Value_new'] = df['customer_count'].astype(float)
df.head()
```

```python
df['monthly_sales_new'] = df['monthly_sales'].fillna(df['monthly_sales'].mean()).astype(int) ## first assigning missing value then changing its type
df.head()
```

```python
df['Increased CTC']=df['CTC'].apply(lambda x:(x+(x/10))) #created a new column with 10% increased CTC
df.head()
```

## Data Aggregating and Grouping


```python
df.head()
```

```python
# Group all them based on age and get mean of there project completed
grouped_mean = df.groupby('age')['projects_completed'].mean()
print(grouped_mean)
```

```python
df.head()
```

```python
grouped_sum = df.groupby(['age', 'score'])['monthly_sales'].sum()
print(grouped_sum)
```

```python
## aggregate multiple function

grouped_agg = df.groupby('score')['age'].agg(['mean', 'sum', 'count'])
print(grouped_agg.head())
```

## Merging and Joining Dataframes

```python
df1 = pd.DataFrame({'key':['A','B','C'], 'value1':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'], 'value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
## Merge dataFrame on the 'Key column'

pd.merge(df1,df2, on='key', how='inner') #merge on basis of key that intersect
```

```python
pd.merge(df1,df2, on='key', how='outer')
```

```python
pd.merge(df1,df2, on='key', how='left')
```

```python
pd.merge(df1,df2, on='key', how='right')
```

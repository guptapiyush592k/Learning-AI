```python
## Handling missing value
df.isnull()
```

```python
df.isnull().any(axis=1)  # it means row wise tell where data is missing
```

```python
df.isnull().any() #it will tell which column has missing value
```

```python
df.isnull().sum() #is telling which column has how many missing value
```

```python
df_filled = df.fillna(0) #replace all the missing value with 0
df_filled
```

```python
## filling missing value with the mean of the column

df['sales_fillNA'] = df['monthly_sales'].fillna(df['monthly_sales'].mean())
df.head(10)
```

```python
## Renaming columns
df = df.rename(columns={'salary':'CTC'})
df.head()
```

```python
##Change dataTypes

df['Value_new'] = df['customer_count'].astype(float)
df.head()
```

```python
df['monthly_sales_new'] = df['monthly_sales'].fillna(df['monthly_sales'].mean()).astype(int) ## first assigning missing value then changing its type
df.head()
```

```python
df['Increased CTC']=df['CTC'].apply(lambda x:(x+(x/10))) #created a new column with 10% increased CTC
df.head()
```

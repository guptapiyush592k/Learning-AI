```python
def main_welcome(func, lst):
    msg='welcome'

    def sub_welcome():
        print("welcome again")
        print(func(lst))
        print('And here we go!!')

    return sub_welcome()
```

```python
## Decorator

def main_welcome(func):
    msg='welcome'

    def sub_welcome():
        print("welcome again")
        func()
        print('And here we go!!')

    return sub_welcome()
```

```python
def course_intro():
    print("Here we are learning decorators")
```

```python
main_welcome(course_intro)
```

```python
@main_welcome
def course_intro():
    print("Here we are learning decorators")
```

```python
## Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called..")
        func()
        print("Something is happening after the function is called..")
    return wrapper
```

```python
@my_decorator
def say_Hello():
    print("Hello")

```

```python
say_Hello()
```

```python
##Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **krgs):
            for _ in range(n):
                func(*args, **krgs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_Hello():
    print("Hello")
```

```python
say_Hello()
```

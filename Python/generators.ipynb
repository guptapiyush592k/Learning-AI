#### Generators

It is used to get value on the fly without consuming any space

```python
def square(n):
    for i in range(3):
        return i ** 2
    
square(3)
```

```python
def square(n):
    for i in range(3):
        yield  i ** 2
    
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
## Alternate to above code

a = square(3)
print(a)
```

```python
next(a)
```

```python
## Another Example
def my_generator():
    yield 1
    yield 2
    yield 3

gen = my_generator()
```

```python
next(gen)
```

```python
### Practical Example -> Reading large file

def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
